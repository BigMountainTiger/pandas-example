```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'id': [1, 2, 3, None, None, None],
    'score': [20, 20, 100, 10, 10, 10]
}).convert_dtypes(dtype_backend='pyarrow')

df
```

```python
condition = df['id'].isnull()
condition
```

```python
df.loc[condition, :]
```

<h1>~ is the negate operator</h1>

```python
df.loc[(~condition), :]
```
