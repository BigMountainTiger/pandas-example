```python
import pandas as pd
```

<h1>1. Student dateframe</h1>

```python
df_student = pd.DataFrame({
    'id': [1, 2, 3, 4, 5, 6],
    'science_score': [None, 0, 0, 0, 0, 0],
    'english_score': [None, 0, 0, 0, 0, 0]
}).convert_dtypes(dtype_backend='pyarrow')

df_student
```

<h1>2. Score dataframe</h1>

```python
df_score = pd.DataFrame({
    'id': [4, 5, 6],
    'science_score': [40, 50, 60],
    'english_score': [44, 55, 66]
}).convert_dtypes(dtype_backend='pyarrow')

df_score
```

<h1>3. merge two dataframes with indicator=True</h1>

```python
df_merged = df_student.merge(df_score, left_on=['id'], right_on=['id'], how='left', suffixes=('', '_r'), indicator=True)
df_merged = df_merged.convert_dtypes(dtype_backend='pyarrow')
df_merged
```

<h1>4. Update the merged rows if df_merged['_merge'] == 'both'</h1>
<h4>Multiple columns</h4>

```python
# df_merged['science_score'] = df_merged['science_score'].astype('int64[pyarrow]')
# df_merged['english_score'] = df_merged['english_score'].astype('int64[pyarrow]')

condition = df_merged['_merge'] == 'both'
update_columns = ['science_score', 'english_score']
source_columns = [f'{i}_r' for i in update_columns]

df_merged.loc[condition, update_columns] = df_merged.loc[condition, source_columns].to_numpy()
df_merged
```
