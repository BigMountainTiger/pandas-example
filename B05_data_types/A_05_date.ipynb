```python
import pandas as pd
from datetime import date, datetime
```

```python
df = pd.DataFrame({
    'id': [1, 2, 3, 4],
    'date': [date(2025, 1, 1), date(2025, 1, 2), date(2025, 1, 3), None]
}).convert_dtypes(dtype_backend='pyarrow')

# For some reason convert_dtypes keeps the date column as object
df.info()
df
```

<h1>The date comparison works fine with object type</h1>

```python
df.loc[(df['date'] <= date(2025, 1, 2)) | df['date'].isnull(), :]
```

```python
# For some reason convert_dtypes keeps the date column as object
df['date'] = df['date'].astype('date32[pyarrow]')
df.info()
df
```

<h1>The date comparison works fine with date32[pyarrow] type as well</h1>

```python
df.loc[(df['date'] <= date(2025, 1, 2)) | df['date'].isnull(), :]
```

<h1>It is a run-time error to compare datetime.datetime to datetime.date</h1>

```python
now = datetime.now()
d = date.fromisoformat(now.strftime('%Y-%m-%d'))

if now > d:
    print('fine')
```
