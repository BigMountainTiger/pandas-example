<h1>1. By default, pandas create an index based range index</h1>

```python
import pandas as pd

data = {
    'name': ['Student 1', 'Student 2', 'Student 3'],
    'score': [71, 72, 73],
}

df = pd.DataFrame(data).convert_dtypes(dtype_backend='pyarrow')

df
```

<h1>2. set_index()</h1>
<h4>set_index() is not repeatable, need to make sure the dataframe has a range index (default index) before calling it</h4>

```python
df = pd.DataFrame(data).convert_dtypes(dtype_backend='pyarrow')
df = df.set_index('name')
df
```

<h1>3. reset_index()</h1>
<h4>reset_index() is not repeatable, need to make sure the dataframe has a non-default index before calling it</h4>

```python
df = pd.DataFrame(data).convert_dtypes(dtype_backend='pyarrow').set_index('name')
df = df.reset_index()
df
```

<h1>4. Once become the index, the column is no longer accessible by name</h1>
<h4>Need to access it by df.index</h4>

```python
df = pd.DataFrame(data).convert_dtypes(dtype_backend='pyarrow')
df = df.set_index('name')

df.loc[df.index == 'Student 1', :]

# The following (access by name) is a run-time error
# df.loc[~df['name'].isnull, :]
```

<h1>5. Duplicate index values are possible</h1>

```python
df_1 = pd.DataFrame({
    'name': ['Student 1', 'Student 2', 'Student 3'],
    'score': [71, 72, 73],
}).convert_dtypes(dtype_backend='pyarrow')

df_2 = pd.DataFrame({
    'name': ['Student 4', 'Student 5', 'Student 6'],
    'score': [74, 75, 76],
}).convert_dtypes(dtype_backend='pyarrow')

df = pd.concat([df_1, df_2])
df
```

<h1>6. Duplicate indices do not affect .loc[] result</h1>

```python
df_1 = pd.DataFrame({
    'name': ['Student 1', 'Student 2', 'Student 3'],
    'score': [71, 72, 73],
}).convert_dtypes(dtype_backend='pyarrow')

df_2 = pd.DataFrame({
    'name': ['Student 4', 'Student 5', 'Student 6'],
    'score': [74, 75, 76],
}).convert_dtypes(dtype_backend='pyarrow')

df = pd.concat([df_1, df_2])
df.loc[df['name'].str.lower() == 'student 3', :]
```
