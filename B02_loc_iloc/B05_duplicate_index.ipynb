```python
import pandas as pd

df_1 = pd.DataFrame({
    'id': [1, 2, 3],
    'name': ['Student 1', 'Student 2', 'Student 3']
}).convert_dtypes(dtype_backend='pyarrow')

df_2 = pd.DataFrame({
    'id': [4, 5, 6],
    'name': ['Student 4', 'Student 5', 'Student 6']
}).convert_dtypes(dtype_backend='pyarrow')

df = pd.concat([df_1, df_2])
df
```

<h1>Duplicate indices do not affect the result of loc[]</h1>

```python
df.loc[df['id'].isin([1, 2, 3]), :]
```
